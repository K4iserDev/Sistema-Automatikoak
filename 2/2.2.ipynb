```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = sns.load_dataset("titanic")
df.head() # 891 errenkada × 15 zutabe
# edo fitxategia igo eta df = pd.read_csv("titanic.csv")
df.head()
```

```python
# Iruzkina1 -> Zutabe bakoitzaren DATU-MOTA itzultzen du (int64, float64, object, category, bool...)
display(df.dtypes)
# Iruzkina2 -> Zutabe bakoitzean zenbat balio DESBERDIN dauden zenbatzen du (nunique)
#              eta emaitza txikienetik handienera ordenatzen du (sort_values)
df.nunique().sort_values()
```

```python
# Zer egiten du hurrengo kodea?
# Zutabe bakoitzaren laburpen-taula bat sortzen du (DataFrame bat), lau zutaberekin:
#   - dtype          -> zutabearen datu-mota, testu gisa (astype(str) dtype-ei bakarrik aplikatzen zaie)
#   - balio_bakarrak -> zenbat balio desberdin dauden (nunique)
#   - nuluak         -> zenbat NaN dauden (isnull().sum())
#   - %_nuluak       -> NaN-en ehunekoa: isnull().mean() = NaN-en proportzioa, *100 eta 1 hamartarrera biribildua
# Horrela begirada batean ikusten da zein zutabek behar duten garbiketa (adib. deck %77.2 hutsik).
laburpena = pd.DataFrame({
    "dtype": df.dtypes.astype(str),
    "balio_bakarrak": df.nunique(),
    "nuluak": df.isnull().sum(),
    "%_nuluak": (df.isnull().mean() * 100).round(1)
})
laburpena
```

**Legenda:** Kn = kategoriko nominala · Ko = kategoriko ordinala · Kj = kuantitatibo jarraitua · Hl = helburu-aldagaia

| Zutabea | Mota | % nulu | Prestaketa-plana |
|---|---|---:|---|
| deck | Kn | 77.2% | Baztertu (nulu gehiegi) |
| age | Kj | 19.9% | Inputatu (mediana) + Eskalatu |
| embarked | Kn | 0.2% | Inputatu (moda) + Kodetu (one-hot) |
| sex | Kn | 0.0% | Kodetu (one-hot / 0-1) |
| fare | Kj | 0.0% | Outlierrak tratatu + Eskalatu |
| pclass | Ko | 0.0% | Ordinala denez, zenbaki gisa utzi daiteke (1<2<3) |
| survived | Hl | 0.0% | Helburu-aldagaia (y), ez da X-en sartzen |
